```python
import whisper
from google.colab import files
from deep_translator import GoogleTranslator
from google.colab import userdata
import google.generativeai as genai
import ast

# --- Interface Language Setup ---
LANGUAGES = {
    'en': 'English',
    'hi': 'Hindi',
    'de': 'German',
    'es': 'Spanish'
}

def select_language():
    print("Select your interface language / अपनी भाषा चुनें / Wählen Sie Ihre Sprache / Seleccione su idioma:")
    for code, name in LANGUAGES.items():
        print(f"{code}: {name}")
    while True:
        lang = input("Enter language code: ").strip().lower()
        if lang in LANGUAGES:
            return lang
        print("Invalid code. Try again.")

def translate(text, target, source='auto'):
    try:
        return GoogleTranslator(source=source, target=target).translate(text)
    except Exception as e:
        print(f"Translation error: {e}")
        return text

# --- STT + Translation Helper ---
stt_model = whisper.load_model("base")

def get_input(prompt, interface_lang):
    prompt_translated = translate(prompt, interface_lang)
    mode = input(f"{prompt_translated}\n{translate('Type t to type or a to upload audio:', interface_lang)} ").strip().lower()
    if mode == 'a':
        print(translate("Please upload your audio file (wav/mp3/m4a):", interface_lang))
        uploaded = files.upload()
        audio_path = next(iter(uploaded))
        result = stt_model.transcribe(audio_path, fp16=False)
        detected_lang = result['language']
        print(f"{translate('Detected language:', interface_lang)} {LANGUAGES.get(detected_lang, detected_lang)}")
        print(f"{translate('Transcribed:', interface_lang)} {result['text']}")
        user_text = result['text']
        # Translate to English if not English
        if detected_lang != 'en':
            try:
                user_text = GoogleTranslator(source='auto', target='en').translate(user_text)
                print(f"{translate('Translated to English:', interface_lang)} {user_text}")
            except Exception as e:
                print(translate('Translation failed:', interface_lang), e)
        return user_text
    else:
        user_text = input(prompt_translated + " ")
        # Optional: Translate typed input if not in English interface
        if interface_lang != 'en' and user_text.strip():
            try:
                user_text = GoogleTranslator(source='auto', target='en').translate(user_text)
            except Exception as e:
                print(translate('Translation failed:', interface_lang), e)
        return user_text

# --- Gemini Setup ---
api_key = userdata.get('GOOGLE_API_KEY')
genai.configure(api_key=api_key)
model = genai.GenerativeModel(model_name="gemini-1.5-flash-latest")
chat = model.start_chat()

# --- Select interface language ---
user_lang = select_language()

# --- Basic Info Collection ---
basic_info = {}

print(translate("Welcome to the Medical Assistant Chatbot!", user_lang))
questions = {
    "name": "Please enter your full name:",
    "phone": "Please enter your phone number:",
    "age": "Please enter your age:",
}

for key, question in questions.items():
    user_input = get_input(question, user_lang)
    basic_info[key] = user_input

# --- Gender Validation with Multilingual Mapping (NO TRANSLATION) ---
gender_map = {
    "male": "Male", "man": "Male", "hombre": "Male", "männlich": "Male", "पुरुष": "Male",
    "female": "Female", "woman": "Female", "mujer": "Female", "weiblich": "Female", "महिला": "Female",
    "other": "Other", "otro": "Other", "andere": "Other", "अन्य": "Other"
}
gender_prompt = {
    'en': "Please specify your gender (male/female/other):",
    'hi': "कृपया अपना लिंग निर्दिष्ट करें (पुरुष/महिला/अन्य):",
    'de': "Bitte geben Sie Ihr Geschlecht an (männlich/weiblich/andere):",
    'es': "Especifique su género (hombre/mujer/otro):"
}
gender_error = {
    'en': "Invalid input. Please say or type male, female, or other.",
    'hi': "अमान्य इनपुट। कृपया पुरुष, महिला या अन्य बोलें या लिखें।",
    'de': "Ungültige Eingabe. Bitte sagen oder schreiben Sie männlich, weiblich oder andere.",
    'es': "Entrada no válida. Por favor diga o escriba hombre, mujer u otro."
}

def get_gender_input(prompt, interface_lang):
    prompt_translated = translate(prompt, interface_lang)
    mode = input(f"{prompt_translated}\n{translate('Type t to type or a to upload audio:', interface_lang)} ").strip().lower()
    if mode == 'a':
        print(translate("Please upload your audio file (wav/mp3/m4a):", interface_lang))
        uploaded = files.upload()
        audio_path = next(iter(uploaded))
        result = stt_model.transcribe(audio_path, fp16=False)
        detected_lang = result['language']
        print(f"{translate('Detected language:', interface_lang)} {LANGUAGES.get(detected_lang, detected_lang)}")
        print(f"{translate('Transcribed:', interface_lang)} {result['text']}")
        user_text = result['text'].strip().lower()
        # DO NOT translate audio input for gender!
        return user_text
    else:
        user_text = input(prompt_translated + " ").strip().lower()
        # DO NOT translate typed input for gender!
        return user_text

while True:
    gender_input = get_gender_input(gender_prompt[user_lang], user_lang)
    mapped_gender = gender_map.get(gender_input)
    if mapped_gender:
        basic_info['gender'] = mapped_gender
        break
    else:
        print(gender_error[user_lang])

# --- Symptom Collection ---
symptom_prompt = "Please describe your symptoms in detail. Mention everything bothering you."
symptoms_input = get_input(symptom_prompt, user_lang)

# --- Extract Symptoms using Gemini ---
extract_prompt = f"""
Extract all orthopedic-related symptoms mentioned in the following sentence:

"{symptoms_input}"

Return ONLY a clean Python list of symptoms. Do NOT include any explanations, code, or additional text.
Example format: ['Symptom1', 'Symptom2']
If no valid orthopedic symptom, return an empty list: []
"""

response = chat.send_message(extract_prompt)
print("\n" + translate("Detected Symptoms:", user_lang))
print(response.text)

# Parse symptoms
try:
    detected_symptoms = ast.literal_eval(response.text.strip())
except:
    detected_symptoms = []

# --- Handle Ambiguous / No Symptoms ---
if not detected_symptoms:
    clarification_prompt = "You mentioned unclear or no symptoms. Could you please specify if you have any orthopedic symptoms such as pain, swelling, stiffness, or injury?"
    symptoms_input = get_input(clarification_prompt, user_lang)

    # Re-run extraction
    extract_prompt = f"""
    Extract all orthopedic-related symptoms mentioned in the following sentence:

    "{symptoms_input}"

    Return ONLY a clean Python list of symptoms. Do NOT include any explanations, code, or additional text.
    Example format: ['Symptom1', 'Symptom2']
    If no valid orthopedic symptom, return an empty list: []
    """
    response = chat.send_message(extract_prompt)
    print("\n" + translate("Detected Symptoms:", user_lang))
    print(response.text)

    try:
        detected_symptoms = ast.literal_eval(response.text.strip())
    except:
        detected_symptoms = []

# --- Follow-up Questions (Only Top 5) ---
symptom_details = {}

if detected_symptoms:
    for symptom in detected_symptoms:
        print(f"\n{translate('Regarding', user_lang)} '{symptom}':")
        location = get_input(f"Where is the {symptom} located?", user_lang)
        duration = get_input(f"How long have you had this {symptom}?", user_lang)
        severity = get_input(f"On a scale of 1 to 10, how severe is your {symptom}?", user_lang)
        pain_type = get_input(f"Would you describe the {symptom} as sharp, dull, aching, burning, etc.?", user_lang)
        assoc_symptom = get_input(f"Any other symptoms associated with {symptom}?", user_lang)

        symptom_details[symptom] = {
            "Location": location,
            "Duration": duration,
            "Severity": severity,
            "Type": pain_type,
            "Associated Symptoms": assoc_symptom
        }

# --- Past Medical Conditions ---
medical_history = get_input("Do you have any ongoing or past medical conditions? Please mention them.", user_lang)

# --- Format Symptom Details ---
symptom_text = ""
for symptom, details in symptom_details.items():
    symptom_text += f"""\n{symptom}:
Location: {details['Location']}
Duration: {details['Duration']}
Severity: {details['Severity']}
Type: {details['Type']}
Associated Symptoms: {details['Associated Symptoms']}\n"""

# --- Summary Generation ---
if detected_symptoms:
    summary_prompt = f"""
Generate a structured orthopedic patient summary report in the following format:

{symptom_text}

Overall Summary:
The patient is {basic_info['name']}, aged {basic_info['age']}, gender {basic_info['gender']}, phone number {basic_info['phone']}.
Medical History: {medical_history}.
"""
else:
    summary_prompt = f"""
Generate a brief patient summary:

The patient is {basic_info['name']}, aged {basic_info['age']}, gender {basic_info['gender']}, phone number {basic_info['phone']}.
No clear orthopedic symptoms reported.
Medical History: {medical_history}.
"""

final_response = chat.send_message(summary_prompt)

# --- Output Summary in User's Language ---
print("\n--- " + translate("Patient Summary Report", user_lang) + " ---\n")
print(translate(final_response.text, user_lang))
```

```python
def chatbot():
    import google.generativeai as genai
    import ast
    import json
    import os
    from datetime import datetime
    import whisper
    from google.colab import files
    from deep_translator import GoogleTranslator
    from gtts import gTTS
    import IPython.display as ipd
    from fpdf import FPDF

    # --- Language Setup ---
    LANGUAGES = {
        'en': 'English',
        'hi': 'Hindi',
        'de': 'German',
        'es': 'Spanish'
    }

    def speak(text, user_lang):
        try:
            tts = gTTS(text=text, lang=user_lang)
            tts.save("response.mp3")
            ipd.display(ipd.Audio("response.mp3", autoplay=True))
            os.remove("response.mp3")
        except Exception as e:
            print(f"TTS Error: {e}")

    def select_language():
        print("Select your interface language / अपनी भाषा चुनें / Wählen Sie Ihre Sprache / Seleccione su idioma:")
        for code, name in LANGUAGES.items():
            print(f"{code}: {name}")
        while True:
            lang = input("Enter language code: ").strip().lower()
            if lang in LANGUAGES:
              speak(f"You selected {LANGUAGES[lang]}", lang)
              return lang
            print("Invalid code. Try again.")

    def translate(text, target, source='auto'):
        try:
            return GoogleTranslator(source=source, target=target).translate(text)
        except Exception as e:
            print(f"Translation error: {e}")
            return text

    # --- Whisper STT ---
    stt_model = whisper.load_model("base")

    def get_audio_input(prompt, interface_lang):
        prompt_translated = translate(prompt, interface_lang)
        print(prompt_translated)
        speak(prompt_translated, interface_lang)
        upload_prompt = translate("Please upload your audio file (wav/mp3/m4a):", interface_lang)
        print(upload_prompt)
        speak(upload_prompt, interface_lang)
        uploaded = files.upload()
        audio_path = next(iter(uploaded))
        result = stt_model.transcribe(audio_path)
        detected_lang = result["language"]
        print(f"Detected language: {LANGUAGES.get(detected_lang, detected_lang)}")
        user_text = result["text"]
        print(f"Transcribed: {user_text}")
        if detected_lang != 'en':
            user_text_en = translate(user_text, 'en', detected_lang)
        else:
            user_text_en = user_text
        return user_text_en, detected_lang, user_text

    def get_text_input(prompt, interface_lang):
        prompt_translated = translate(prompt, interface_lang)
        print(prompt_translated)
        speak(prompt_translated, interface_lang)
        user_text = input()
        if interface_lang != 'en':
            user_text_en = translate(user_text, 'en', interface_lang)
        else:
            user_text_en = user_text
        return user_text_en, interface_lang, user_text

    def get_input(prompt, interface_lang):
        prompt_mode = translate("Type 't' to type or 'a' to upload audio: ", interface_lang)
        speak(prompt_mode, interface_lang)
        while True:
            mode = input(prompt_mode).strip().lower()
            if mode in ['t', 'a']:
                break
            error_text = translate("Invalid input. Please enter 't' or 'a'.", interface_lang)
            print(error_text)
            speak(error_text, interface_lang)
        if mode == 'a':
            return get_audio_input(prompt, interface_lang)
        else:
            return get_text_input(prompt, interface_lang)

    # --- Main Chatbot Logic ---
    user_lang = select_language()

    from google.colab import userdata
    api_key = userdata.get('GOOGLE_API_KEY')
    genai.configure(api_key=api_key)
    model = genai.GenerativeModel(model_name="gemini-1.5-flash-latest")
    chat = model.start_chat()

    basic_info = {}
    questions = {
        "name": "Please enter your full name:",
        "phone": "Please enter your phone number:",
        "age": "Please enter your age:",
    }



    # --- Gender Input (no translation, robust mapping) ---
    gender_map = {
        "male": "Male", "man": "Male", "hombre": "Male", "männlich": "Male", "पुरुष": "Male",
        "female": "Female", "woman": "Female", "mujer": "Female", "weiblich": "Female", "महिला": "Female",
        "other": "Other", "otro": "Other", "andere": "Other", "अन्य": "Other"
    }
    gender_prompt = {
        'en': "Please specify your gender (male/female/other):",
        'hi': "कृपया अपना लिंग निर्दिष्ट करें (पुरुष/महिला/अन्य):",
        'de': "Bitte geben Sie Ihr Geschlecht an (männlich/weiblich/andere):",
        'es': "Especifique su género (hombre/mujer/otro):"
    }
    gender_error = {
        'en': "Invalid input. Please say or type male, female, or other.",
        'hi': "अमान्य इनपुट। कृपया पुरुष, महिला या अन्य बोलें या लिखें।",
        'de': "Ungültige Eingabe. Bitte sagen oder schreiben Sie männlich, weiblich oder andere.",
        'es': "Entrada no válida. Por favor diga o escriba hombre, mujer u otro."
    }

    # --- Basic Info Collection ---
    basic_info = {}
    print(translate("Welcome to the Medical Assistant Chatbot!", user_lang))
    questions = {
        "name": "Please enter your full name:",
        "phone": "Please enter your phone number:",
        "age": "Please enter your age:",
    }
    for key, question in questions.items():
        user_input = get_input(question, user_lang)
        basic_info[key] = user_input

    # --- Gender Validation ---
    while True:
        user_text_en, detected_lang, user_text_orig = get_input(gender_prompt[user_lang], user_lang)
        valid = False
        user_input_clean = user_text_orig.strip().lower()
        if user_input_clean in gender_map:
            basic_info['gender'] = gender_map[user_input_clean]
            valid = True
            break
        if not valid:
          if user_text_en.strip().lower() in gender_map['en']:
                basic_info['gender'] = gender_map['en'][user_text_en.strip().lower()]
                valid = True
        if valid:
            break
        else:
            print(gender_error[user_lang])
            speak(gender_error[user_lang], user_lang)

    # --- Symptom Collection ---
    symptom_prompt = "Please describe your symptoms in detail. Mention everything bothering you."
    symptoms_input, _, _ = get_input(symptom_prompt, user_lang)


    # --- Extract Symptoms using Gemini ---
    extract_prompt = f"""
    Extract all orthopedic-related symptoms mentioned in the following sentence:

    "{symptoms_input}"

    Return ONLY a clean Python list of symptoms. Do NOT include any explanations, code, or additional text.
    Example format: ['Symptom1', 'Symptom2']
    If no valid orthopedic symptom, return an empty list: []
    """
    response = chat.send_message(extract_prompt)
    print("\n" + translate("Detected Symptoms:", user_lang))
    speak(translate("Detected Symptoms:", user_lang), user_lang)
    print(response.text)
    speak(translate(response.text, user_lang), user_lang)

    try:
        detected_symptoms = ast.literal_eval(response.text.strip())
    except:
        detected_symptoms = []

    # --- Handle Ambiguous / No Symptoms ---
    if not detected_symptoms:
        clarification_prompt = "You mentioned unclear or no symptoms. Could you please specify if you have any orthopedic symptoms such as pain, swelling, stiffness, or injury?"
        symptoms_input = get_input(clarification_prompt, user_lang)
        extract_prompt = f"""
        Extract all orthopedic-related symptoms mentioned in the following sentence:

        "{symptoms_input}"

        Return ONLY a clean Python list of symptoms. Do NOT include any explanations, code, or additional text.
        Example format: ['Symptom1', 'Symptom2']
        If no valid orthopedic symptom, return an empty list: []
        """
        response = chat.send_message(extract_prompt)
        print("\n" + translate("Detected Symptoms:", user_lang))
        speak(translate("Detected Symptoms:", user_lang), user_lang)
        print(response.text)
        speak(translate(response.text, user_lang), user_lang)
        try:
            detected_symptoms = ast.literal_eval(response.text.strip())
        except:
            detected_symptoms = []

    # --- Follow-up Questions (Only Top 5) ---
    symptom_details = {}
    if detected_symptoms:
        for symptom in detected_symptoms:
            print(f"\n{translate('Regarding', user_lang)} '{symptom}':")
            speak(f"{translate('Regarding', user_lang)} {symptom}", user_lang)
            location = get_input(f"Where is the {symptom} located?", user_lang)
            duration = get_input(f"How long have you had this {symptom}?", user_lang)
            severity = get_input(f"On a scale of 1 to 10, how severe is your {symptom}?", user_lang)
            pain_type = get_input(f"Would you describe the {symptom} as sharp, dull, aching, burning, etc.?", user_lang)
            assoc_symptom = get_input(f"Any other symptoms associated with {symptom}?", user_lang)
            symptom_details[symptom] = {
                "Location": location,
                "Duration": duration,
                "Severity": severity,
                "Type": pain_type,
                "Associated Symptoms": assoc_symptom
            }

    # --- Past Medical Conditions ---
    medical_history,_,t_ = get_input("Do you have any ongoing or past medical conditions? Please mention them.", user_lang)
    #print(medical_history)
    #print(type(medical_history))
    medical_condition_details = ""
    if medical_history.strip().lower() != "no" and len(medical_history.strip()) > 3:
        followup_prompt = f"""
    You are a medical assistant. The patient mentioned the following medical condition(s): "{medical_history}".

    Generate 3-4 concise follow-up questions to gather more details about these condition(s). Keep the questions relevant, specific, and easy to understand. Do NOT include explanations.
    Just list the questions.
    """
        followup_response = chat.send_message(followup_prompt)
        print("\n" + translate("Follow-up questions regarding your medical condition(s):", user_lang) + "\n")
        print(followup_response.text)
        followup_questions = followup_response.text.strip().split("\n")
        followup_answers = []
        for q in followup_questions:
            answer = get_input(q.strip(), user_lang)
            followup_answers.append({
                "question": q.strip(),
                "answer": answer
            })
        medical_condition_details = "\nAdditional Details:\n"
        for item in followup_answers:
            medical_condition_details += f"{item['question']} {item['answer']}\n"

    # --- Format Symptom Details ---
    symptom_text = ""
    for symptom, details in symptom_details.items():
        symptom_text += f"""\n{symptom}:
    Location: {details['Location']}
    Duration: {details['Duration']}
    Severity: {details['Severity']}
    Type: {details['Type']}
    Associated Symptoms: {details['Associated Symptoms']}\n"""

    # --- Summary Generation ---
    if detected_symptoms:
        summary_prompt = f"""
    Generate a structured orthopedic patient summary text report about 60-80 words using the following info the report should be concise as well as informative:

    **Patient Information:**

    * Name: {basic_info['name']}
    * Age: {basic_info['age']}
    * Gender: {basic_info['gender']}
    * Phone Number: {basic_info['phone']}

    **Chief Complaint:** {', '.join(detected_symptoms)}

    **Detailed Complaint:**
    {symptom_text}

    **Medical History:** {medical_history}
    {medical_condition_details}

    **Overall Summary:**
    """
    else:
        summary_prompt = f"""
    Generate a brief patient summary:

    **Patient Information:**

    * Name: {basic_info['name']}
    * Age: {basic_info['age']}
    * Gender: {basic_info['gender']}
    * Phone Number: {basic_info['phone']}

    No clear orthopedic symptoms reported.

    **Medical History:** {medical_history}
    {medical_condition_details}

    **Overall Summary:**
    """

    final_response = chat.send_message(summary_prompt)

    # --- Output Summary ---
    print("\n--- " + translate("Patient Summary Report", user_lang) + " ---\n")
    print(translate(final_response.text, user_lang))

    # --- Save Output in JSON Format ---
    patient_json = {
        "user_details": {
            "name": basic_info['name'],
            "phone": basic_info['phone'],
            "age": basic_info['age'],
            "gender": basic_info['gender']
        },
        "symptom_data": [],
        "medical_conditions": {
            "condition": medical_history,
            "follow_up_details": medical_condition_details.strip()
        },
        "overall_summary": final_response.text.strip()
    }
    for symptom, details in symptom_details.items():
        patient_json["symptom_data"].append({
            "symptom": symptom,
            "location": details['Location'],
            "duration": details['Duration'],
            "severity": details['Severity'],
            "type": details['Type'],
            "associated_symptoms": details['Associated Symptoms']
        })
    with open("patient_summary.json", "w") as f:
        json.dump(patient_json, f, indent=4)
    print("\n" + translate("Patient summary saved as", user_lang) + " patient_summary.json")

    # --- Save JSON File ---
    filename = "summary.json"
    with open(filename, 'w') as json_file:
        json.dump(patient_json, json_file, indent=4)
    print(f"\n✅ Patient summary saved successfully as '{filename}'")
    def generate_report(json_filename, output_filename):
        with open(json_filename, 'r') as file:
            data = json.load(file)

        pdf = FPDF()
        pdf.set_auto_page_break(auto=True, margin=15)
        pdf.add_page()

        # Title
        pdf.set_font("Arial", 'B', 18)
        pdf.cell(0, 10, "ORTHOPEDIC PATIENT SCREENING REPORT", ln=True, align='C')
        pdf.ln(10)

        # Patient Details
        pdf.set_font("Arial", 'B', 14)
        pdf.cell(0, 10, "Patient Information", ln=True, border=1)
        pdf.set_font("Arial", size=12)
        pdf.ln(5)

        user_details = data["user_details"]
        pdf.cell(50, 10, "Name:", border=1)
        pdf.cell(140, 10, user_details.get("name", ""), border=1, ln=True)

        pdf.cell(50, 10, "Age:", border=1)
        pdf.cell(140, 10, str(user_details.get("age", "")), border=1, ln=True)

        pdf.cell(50, 10, "Gender:", border=1)
        pdf.cell(140, 10, user_details.get("gender", ""), border=1, ln=True)

        pdf.cell(50, 10, "Phone Number:", border=1)
        pdf.cell(140, 10, user_details.get("phone", ""), border=1, ln=True)

        pdf.ln(10)

        # Symptoms Summary
        pdf.set_font("Arial", 'B', 14)
        pdf.cell(0, 10, "Symptoms Summary", ln=True, border=1)
        pdf.set_font("Arial", size=12)
        pdf.ln(5)

        headers = ["Symptom", "Location", "Duration", "Severity", "Type", "Associated Symptoms"]
        col_widths = [30, 30, 30, 20, 40, 40]
        row_height = 10

        # Headers
        for i, header in enumerate(headers):
            pdf.cell(col_widths[i], row_height, header, border=1, align='C')
        pdf.ln(row_height)

        # Data Rows
        for symptom_entry in data["symptom_data"]:
            pdf.cell(col_widths[0], row_height, symptom_entry.get("symptom", ""), border=1)
            pdf.cell(col_widths[1], row_height, symptom_entry.get("location", ""), border=1)
            pdf.cell(col_widths[2], row_height, symptom_entry.get("duration", ""), border=1)
            pdf.cell(col_widths[3], row_height, symptom_entry.get("severity", ""), border=1)
            pdf.cell(col_widths[4], row_height, symptom_entry.get("type", ""), border=1)
            pdf.cell(col_widths[5], row_height, symptom_entry.get("associated_symptoms", ""), border=1, ln=True)

        pdf.ln(10)

        # Medical Conditions
        pdf.set_font("Arial", 'B', 14)
        pdf.cell(0, 10, "Ongoing/Past Medical Conditions", ln=True, border=1)
        pdf.set_font("Arial", size=12)
        pdf.ln(5)
        medical_conditions = data.get("medical_conditions", {})
        condition_text = f"Condition: {medical_conditions.get('condition', 'None')}\nFollow-up Details: {medical_conditions.get('follow_up_details', 'None')}"
        pdf.multi_cell(0, 10, condition_text)
        pdf.ln(10)

        # Overall Summary
        # Overall Summary
        pdf.set_font("Arial", 'B', 14)
        pdf.cell(0, 10, "Overall Summary", ln=True, border=1)
        pdf.set_font("Arial", size=12)
        overall_summary = data.get("overall_summary", "")
        if isinstance(overall_summary, list):
          overall_summary = "\n".join(overall_summary)
        pdf.multi_cell(0, 10, overall_summary)


    generate_report(filename,"output.pdf")

chatbot()
```

```python
import time
def chatbot():
    try:
        import google.generativeai as genai
        import ast
        import json
        import os
        from datetime import datetime
        import whisper
        from google.colab import files
        from deep_translator import GoogleTranslator
        from gtts import gTTS
        import IPython.display as ipd
        from fpdf import FPDF
        import sys

        # --- Language Setup ---
        LANGUAGES = {
            'en': 'English',
            'hi': 'Hindi',
            'de': 'German',
            'es': 'Spanish'
        }

        def speak(text, user_lang):
            try:
                tts = gTTS(text=text, lang=user_lang)
                tts.save("response.mp3")
                ipd.display(ipd.Audio("response.mp3", autoplay=True))
                os.remove("response.mp3")
            except Exception as e:
                print(f"TTS Error: {e}")

        def select_language():
            print("Select your interface language / अपनी भाषा चुनें / Wählen Sie Ihre Sprache / Seleccione su idioma:")
            for code, name in LANGUAGES.items():
                print(f"{code}: {name}")
            while True:
                try:
                    lang = input("Enter language code: ").strip().lower()
                    if lang in LANGUAGES:
                        speak(f"You selected {LANGUAGES[lang]}", lang)
                        return lang
                    print("Invalid code. Try again.")
                except KeyboardInterrupt:
                    print("\nOperation cancelled. Exiting...")
                    sys.exit(0)

        def translate(text, target, source='auto'):
            try:
                return GoogleTranslator(source=source, target=target).translate(text)
            except Exception as e:
                print(f"Translation error: {e}")
                return text

        # --- Whisper STT ---
        stt_model = whisper.load_model("base")

        def get_audio_input(prompt, interface_lang):
            try:
                prompt_translated = translate(prompt, interface_lang)
                print(prompt_translated)
                speak(prompt_translated, interface_lang)
                upload_prompt = translate("Please upload your audio file (wav/mp3/m4a):", interface_lang)
                print(upload_prompt)
                speak(upload_prompt, interface_lang)
                uploaded = files.upload()
                audio_path = next(iter(uploaded))
                result = stt_model.transcribe(audio_path)
                detected_lang = result["language"]
                print(f"Detected language: {LANGUAGES.get(detected_lang, detected_lang)}")
                user_text = result["text"]
                print(f"Transcribed: {user_text}")
                if detected_lang != 'en':
                    user_text_en = translate(user_text, 'en', detected_lang)
                else:
                    user_text_en = user_text
                return user_text_en, detected_lang, user_text
            except KeyboardInterrupt:
                raise
            except Exception as e:
                print(f"Audio input error: {e}")
                return "", interface_lang, ""

        def get_text_input(prompt, interface_lang):
            try:
                prompt_translated = translate(prompt, interface_lang)
                print(prompt_translated)
                speak(prompt_translated, interface_lang)
                user_text = input()
                if interface_lang != 'en':
                    user_text_en = translate(user_text, 'en', interface_lang)
                else:
                    user_text_en = user_text
                return user_text_en, interface_lang, user_text
            except KeyboardInterrupt:
                raise

        def get_input(prompt, interface_lang):
            try:
                prompt_mode = translate("Type 't' to type or 'a' to upload audio : ", interface_lang)
                speak(prompt_mode, interface_lang)
                while True:
                    try:
                        mode = input(prompt_mode).strip().lower()
                        if mode in ['t', 'a', 'l']:
                            break
                        error_text = translate("Invalid input. Please enter 't', 'a', or 'l'.", interface_lang)
                        print(error_text)
                        speak(error_text, interface_lang)
                    except KeyboardInterrupt:
                        raise

                if mode == 'l':
                    new_lang = select_language()
                    return get_input(prompt, new_lang)
                elif mode == 'a':
                    return get_audio_input(prompt, interface_lang)
                else:
                    return get_text_input(prompt, interface_lang)
            except KeyboardInterrupt:
                print(translate("\nOperation cancelled.", interface_lang))
                raise

        # --- Main Chatbot Logic ---
        try:
            user_lang = select_language()

            # Handle API key
            try:
                from google.colab import userdata
                api_key = userdata.get('GOOGLE_API_KEY')
            except (ImportError, userdata.SecretNotFoundError):
                import getpass
                api_key = getpass.getpass("Enter your Google API key: ")

            genai.configure(api_key=api_key)
            model = genai.GenerativeModel(model_name="gemini-1.5-flash-latest")
            chat = model.start_chat()

            # --- Gender Input Handling ---
            gender_map = {
                "male": "Male", "man": "Male", "hombre": "Male", "männlich": "Male", "पुरुष": "Male",
                "female": "Female", "woman": "Female", "mujer": "Female", "weiblich": "Female", "महिला": "Female",
                "other": "Other", "otro": "Other", "andere": "Other", "अन्य": "Other"
            }
            gender_prompt = {
                'en': "Please specify your gender (male/female/other):",
                'hi': "कृपया अपना लिंग निर्दिष्ट करें (पुरुष/महिला/अन्य):",
                'de': "Bitte geben Sie Ihr Geschlecht an (männlich/weiblich/andere):",
                'es': "Especifique su género (hombre/mujer/otro):"
            }
            gender_error = {
                'en': "Invalid input. Please say or type male, female, or other.",
                'hi': "अमान्य इनपुट। कृपया पुरुष, महिला या अन्य बोलें या लिखें।",
                'de': "Ungültige Eingabe. Bitte sagen oder schreiben Sie männlich, weiblich oder andere.",
                'es': "Entrada no válida. Por favor diga o escriba hombre, mujer u otro."
            }

            # --- Basic Info Collection ---
            basic_info = {}
            print(translate("Welcome to the Medical Assistant Chatbot!", user_lang))
            questions = {
                "name": "Please enter your full name:",
                "phone": "Please enter your phone number:",
                "age": "Please enter your age:",
            }
            for key, question in questions.items():
                try:
                    user_input = get_input(question, user_lang)
                    basic_info[key] = user_input
                except KeyboardInterrupt:
                    print(translate("\n\nExiting HealthBot. Thank you for using our service!", user_lang))
                    sys.exit(0)

            # --- Gender Validation ---
            while True:
                try:
                    user_text_en, detected_lang, user_text_orig = get_input(gender_prompt[user_lang], user_lang)
                    valid = False
                    user_input_clean = user_text_orig.strip().lower()
                    if user_input_clean in gender_map:
                        basic_info['gender'] = gender_map[user_input_clean]
                        valid = True
                        break
                    if not valid:
                        if user_text_en.strip().lower() in gender_map['en']:
                            basic_info['gender'] = gender_map['en'][user_text_en.strip().lower()]
                            valid = True
                    if valid:
                        break
                    else:
                        print(gender_error[user_lang])
                        speak(gender_error[user_lang], user_lang)
                except KeyboardInterrupt:
                    print(translate("\n\nExiting HealthBot. Thank you for using our service!", user_lang))
                    sys.exit(0)

            # --- Symptom Collection ---
            symptom_prompt = "Please describe your symptoms in detail. Mention everything bothering you."
            try:
                symptoms_input, _, _ = get_input(symptom_prompt, user_lang)
            except KeyboardInterrupt:
                print(translate("\n\nExiting HealthBot. Thank you for using our service!", user_lang))
                sys.exit(0)

            # --- Symptom Extraction ---
            extract_prompt = f"""
            Extract all orthopedic-related symptoms mentioned in the following sentence:

            "{symptoms_input}"

            Return ONLY a clean Python list of symptoms. Do NOT include any explanations, code, or additional text.
            Example format: ['Symptom1', 'Symptom2']
            If no valid orthopedic symptom, return an empty list: []
            """
            response = chat.send_message(extract_prompt)
            print("\n" + translate("Detected Symptoms:", user_lang))
            speak(translate("Detected Symptoms:", user_lang), user_lang)
            time.sleep(1)
            print(response.text)
            speak(translate(response.text, user_lang), user_lang)
            time.sleep(1.5)

            try:
                detected_symptoms = ast.literal_eval(response.text.strip())
            except:
                detected_symptoms = []

            # --- Handle Ambiguous Symptoms ---
            if not detected_symptoms:
                clarification_prompt = "You mentioned unclear or no symptoms. Could you please specify if you have any orthopedic symptoms such as pain, swelling, stiffness, or injury?"
                try:
                    symptoms_input = get_input(clarification_prompt, user_lang)
                except KeyboardInterrupt:
                    print(translate("\n\nExiting HealthBot. Thank you for using our service!", user_lang))
                    sys.exit(0)

                extract_prompt = f"""
                Extract all orthopedic-related symptoms mentioned in the following sentence:

                "{symptoms_input}"

                Return ONLY a clean Python list of symptoms. Do NOT include any explanations, code, or additional text.
                Example format: ['Symptom1', 'Symptom2']
                If no valid orthopedic symptom, return an empty list: []
                """
                response = chat.send_message(extract_prompt)
                print("\n" + translate("Detected Symptoms:", user_lang))
                speak(translate("Detected Symptoms:", user_lang), user_lang)
                time.sleep(1)
                print(response.text)
                speak(translate(response.text, user_lang), user_lang)
                time.sleep(1)
                try:
                    detected_symptoms = ast.literal_eval(response.text.strip())
                except:
                    detected_symptoms = []

            # --- Symptom Follow-up ---
            symptom_details = {}
            if detected_symptoms:
                for symptom in detected_symptoms:
                    try:
                        print(f"\n{translate('Regarding', user_lang)} '{symptom}':")
                        speak(f"{translate('Regarding', user_lang)} {symptom}", user_lang)
                        time.sleep(1)
                        location = get_input(f"Where is the {symptom} located?", user_lang)
                        user_lang = location[1]  # Update language if changed during input
                        duration = get_input(f"How long have you had this {symptom}?", user_lang)
                        user_lang = duration[1]  # Update language if changed during input
                        severity = get_input(f"On a scale of 1 to 10, how severe is your {symptom}?", user_lang)
                        user_lang = severity[1]  # Update language if changed during input
                        pain_type = get_input(f"Would you describe the {symptom} as sharp, dull, aching, burning, etc.?", user_lang)
                        user_lang = pain_type[1]  # Update language if changed during input
                        assoc_symptom = get_input(f"Any other symptoms associated with {symptom}?", user_lang)
                        user_lang = assoc_symptom[1]  # Update language if changed during input
                        symptom_details[symptom] = {
                            "Location": location,
                            "Duration": duration,
                            "Severity": severity,
                            "Type": pain_type,
                            "Associated Symptoms": assoc_symptom
                        }
                    except KeyboardInterrupt:
                        print(translate("\n\nExiting HealthBot. Thank you for using our service!", user_lang))
                        sys.exit(0)

            # --- Medical History ---
            try:
                medical_history,_,t_ = get_input("Do you have any ongoing or past medical conditions? Please mention them.", user_lang)
            except KeyboardInterrupt:
                print(translate("\n\nExiting HealthBot. Thank you for using our service!", user_lang))
                sys.exit(0)

            medical_condition_details = ""
            if medical_history.strip().lower() != "no" and len(medical_history.strip()) > 3:
                followup_prompt = f"""
                You are a medical assistant. The patient mentioned the following medical condition(s): "{medical_history}".

                Generate 3-4 concise follow-up questions to gather more details about these condition(s). Keep the questions relevant, specific, and easy to understand. Do NOT include explanations.
                Just list the questions.
                """
                followup_response = chat.send_message(followup_prompt)
                print("\n" + translate("Follow-up questions regarding your medical condition(s):", user_lang) + "\n")
                print(followup_response.text)
                followup_questions = followup_response.text.strip().split("\n")
                followup_answers = []
                for q in followup_questions:
                    try:
                        answer = get_input(q.strip(), user_lang)
                        user_lang = answer[1]  # Update language if changed during input
                        followup_answers.append({
                            "question": q.strip(),
                            "answer": answer
                        })
                    except KeyboardInterrupt:
                        print(translate("\n\nExiting HealthBot. Thank you for using our service!", user_lang))
                        sys.exit(0)
                medical_condition_details = "\nAdditional Details:\n"
                for item in followup_answers:
                    medical_condition_details += f"{item['question']} {item['answer']}\n"

            # --- Report Generation ---
            symptom_text = ""
            for symptom, details in symptom_details.items():
                symptom_text += f"""\n{symptom}:
            Location: {details['Location']}
            Duration: {details['Duration']}
            Severity: {details['Severity']}
            Type: {details['Type']}
            Associated Symptoms: {details['Associated Symptoms']}\n"""

            summary_prompt = f"""
            Generate a structured orthopedic patient summary text report about 60-80 words using the following info the report should be concise as well as informative:

            **Patient Information:**

            * Name: {basic_info['name']}
            * Age: {basic_info['age']}
            * Gender: {basic_info['gender']}
            * Phone Number: {basic_info['phone']}

            **Chief Complaint:** {', '.join(detected_symptoms) if detected_symptoms else 'No clear symptoms reported'}

            **Detailed Complaint:**
            {symptom_text if detected_symptoms else 'No detailed symptoms provided'}

            **Medical History:** {medical_history}
            {medical_condition_details}

            **Overall Summary:**
            """

            final_response = chat.send_message(summary_prompt)

            # --- Display Summary ---
            print("\n--- " + translate("Patient Summary Report", user_lang) + " ---\n")
            print(translate(final_response.text, user_lang))

            # --- Save JSON ---
            patient_json = {
                "user_details": basic_info,
                "symptom_data": [{
                    "symptom": symptom,
                    "location": details['Location'],
                    "duration": details['Duration'],
                    "severity": details['Severity'],
                    "type": details['Type'],
                    "associated_symptoms": details['Associated Symptoms']
                } for symptom, details in symptom_details.items()],
                "medical_conditions": {
                    "condition": medical_history,
                    "follow_up_details": medical_condition_details.strip()
                },
                "overall_summary": final_response.text.strip()
            }

            filename = "summary.json"
            with open(filename, 'w') as json_file:
                json.dump(patient_json, json_file, indent=4)
            print(f"\n✅ {translate('Patient summary saved successfully as', user_lang)} '{filename}'")

            # --- PDF Generation ---
            def generate_report(json_filename, output_filename):
                try:
                    with open(json_filename, 'r') as file:
                        data = json.load(file)

                    pdf = FPDF()
                    pdf.set_auto_page_break(auto=True, margin=15)
                    pdf.add_page()

                    # Title
                    pdf.set_font("Arial", 'B', 18)
                    pdf.cell(0, 10, "ORTHOPEDIC PATIENT SCREENING REPORT", ln=True, align='C')
                    pdf.ln(10)
                    # Helper function to clean data
                    def clean_data(value):
                        if isinstance(value, list):
                            # Remove language codes and duplicates
                            cleaned = [item for item in value if not isinstance(item, str) or
                                    (isinstance(item, str) and item not in LANGUAGES.keys())]
                            if cleaned:
                                # Get first unique value if multiple exist
                                seen = set()
                                unique_values = []
                                for item in cleaned:
                                    if item not in seen:
                                        seen.add(item)
                                        unique_values.append(str(item))
                                return " ".join(unique_values)
                            return ""
                        elif isinstance(value, str):
                            return value
                        else:
                            return str(value)

                    # Patient Details
                    pdf.set_font("Arial", 'B', 14)
                    pdf.cell(0, 10, translate("Patient Information", user_lang), ln=True, border=1)
                    pdf.set_font("Arial", size=12)
                    pdf.ln(5)

                    user_details = data["user_details"]
                    details_map = {
                        "name": translate("Name:", user_lang),
                        "age": translate("Age:", user_lang),
                        "gender": translate("Gender:", user_lang),
                        "phone": translate("Phone Number:", user_lang)
                    }

                    for field, label in details_map.items():
                        pdf.cell(50, 10, label, border=1)
                        cell_value = clean_data(user_details.get(field, ""))
                        pdf.cell(140, 10, cell_value, border=1, ln=True)

                    pdf.ln(10)

                    # Symptoms Summary
                    if data["symptom_data"]:
                      pdf.set_font("Arial", 'B', 14)
                      pdf.cell(0, 10, translate("Symptoms Summary", user_lang), ln=True, border=1)
                      pdf.set_font("Arial", size=12)
                      pdf.ln(5)

                      headers = [
                          translate("Symptom", user_lang),
                          translate("Location", user_lang),
                          translate("Duration", user_lang),
                          translate("Severity", user_lang),
                          translate("Type", user_lang),
                          translate("Associated Symptoms", user_lang)
                      ]

                      col_widths = [30, 30, 30, 20, 40, 40]
                      row_height = 10

                      # Headers
                      pdf.set_fill_color(200, 220, 255)
                      for i, header in enumerate(headers):
                          pdf.cell(col_widths[i], row_height, header, border=1, align='C', fill=True)
                      pdf.ln(row_height)

                      # Data Rows
                      pdf.set_fill_color(255, 255, 255)
                      for symptom_entry in data["symptom_data"]:
                          for i, field in enumerate(["symptom", "location", "duration", "severity", "type", "associated_symptoms"]):
                              value = clean_data(symptom_entry.get(field, ""))
                              pdf.cell(col_widths[i], row_height, value, border=1)
                          pdf.ln()

                      pdf.ln(10)
                    # Medical Conditions
                    pdf.set_font("Arial", 'B', 14)
                    pdf.cell(0, 10, translate("Medical History", user_lang), ln=True, border=1)
                    pdf.set_font("Arial", size=12)
                    pdf.ln(5)

                    medical_conditions = data.get("medical_conditions", {})
                    condition_text = (f"{translate('Condition:', user_lang)} {clean_data(medical_conditions.get('condition', 'None'))}\n"
                                    f"{translate('Follow-up Details:', user_lang)} {clean_data(medical_conditions.get('follow_up_details', 'None'))}")
                    pdf.multi_cell(0, 10, condition_text)
                    pdf.ln(10)

                    # Overall Summary
                    pdf.set_font("Arial", 'B', 14)
                    pdf.cell(0, 10, translate("Overall Summary", user_lang), ln=True, border=1)
                    pdf.set_font("Arial", size=12)
                    overall_summary = data.get("overall_summary", "")

                    # Clean the summary text
                    if isinstance(overall_summary, list):
                        overall_summary = clean_data(overall_summary)
                    translated_summary = translate(overall_summary, user_lang)
                    pdf.multi_cell(0, 10, translated_summary)

                    # Save PDF
                    pdf.output(output_filename)
                    print(translate(f"\nPDF report generated: {output_filename}", user_lang))
                    speak(translate("PDF report generated successfully", user_lang), user_lang)

                except KeyboardInterrupt:
                    print(translate("\nReport generation cancelled.", user_lang))
                    speak(translate("Report generation cancelled", user_lang), user_lang)
                    if os.path.exists(output_filename):
                        os.remove(output_filename)
                    sys.exit(0)
                except Exception as e:
                    print(f"PDF Error: {str(e)}")
                    speak(translate("Error generating PDF report", user_lang), user_lang)
                    if os.path.exists(output_filename):
                        os.remove(output_filename)

            generate_report(filename, "output.pdf")

        except KeyboardInterrupt:
            print(translate("\n\nExiting HealthBot. Thank you for using our service!", user_lang if 'user_lang' in locals() else 'en'))
            sys.exit(0)

    except KeyboardInterrupt:
        print("\n\nSession terminated by user. Goodbye!")
        sys.exit(0)

chatbot()
```
